# Gold Trading: Skill or Luck?
Significance test on my 74 real-money XAU/USD trades (MetaTrader 5, Jan–May 2026). P/L in euros, before trading costs.

**Question:** is my +€49 net result evidence of an edge, or could it be luck?

```python
import numpy as np
from scipy import stats

pl = np.array([-9, 15, -8, 13, 11, 10, -9, -25, -6, 11, 13, -4, -7, -6, -7, -9, 15, 12, -6, 15, 13, 14, -9, 11, -8, 12, 18, -9, 13, -5, 17, 16, -6, 18, 13, -9, -30, -4, -5, 10, -6, -7, 14, 11, -5, 15, -5, -9, -7, 16, -9, 17, 12, 14, 12, -5, 18, 18, -6, -9, -40, -7, -8, -7, -6, -5, -5, -8, -7, -4, -4, -4, -5, -9])

print(f'Trades: {len(pl)}  Net: €{pl.sum()}  Win rate: {(pl>0).mean():.1%}')
```

## 1. One-sample t-test
H₀: average P/L per trade = 0. H₁: average P/L > 0.

```python
t = stats.ttest_1samp(pl, 0, alternative='greater')
print(f't = {t.statistic:.2f}, p = {t.pvalue:.2f}')
```

## 2. Bootstrap
Resample the 74 trades 100,000 times to see the plausible range of the total result.

```python
boot = np.random.default_rng(0).choice(pl, (100_000, len(pl))).sum(axis=1)
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f'95% range for net result: €{lo:.0f} to €{hi:.0f}')
print(f'Share of resamples at or below €0: {(boot<=0).mean():.0%}')
```

## Conclusion
p ≈ 0.32, and the 95% range includes €0, so the +€49 edge is **not statistically significant**. A larger sample of trades would be needed to tell skill from luck.
